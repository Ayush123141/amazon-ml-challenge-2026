```python
import pandas as pd
from pathlib import Path

BASE = Path(r"C:\Users\jumbo\Downloads\6ab10eb3b23ba_student_resource\student_resource\dataset")

train_s1 = pd.read_csv(
    BASE / "train/train_source1.tsv",
    sep="\t"
)

train_s2 = pd.read_csv(
    BASE / "train/train_source2.tsv",
    sep="\t"
)

train_s3 = pd.read_csv(
    BASE / "train/train_source3.tsv",
    sep="\t"
)

ground_truth = pd.read_csv(
    BASE / "train/train_ground_truth.tsv",
    sep="\t"
)

test_s1 = pd.read_csv(
    BASE / "test/test_source1.tsv",
    sep="\t"
)

test_s2 = pd.read_csv(
    BASE / "test/test_source2.tsv",
    sep="\t"
)

test_s3 = pd.read_csv(
    BASE / "test/test_source3.tsv",
    sep="\t"
)

print("S1:", train_s1.shape)
print("S2:", train_s2.shape)
print("S3:", train_s3.shape)
print("Ground truth:", ground_truth.shape)

print("Test S1:", test_s1.shape)
print("Test S2:", test_s2.shape)
print("Test S3:", test_s3.shape)
```

```python
datasets = {
    "train_s1": train_s1,
    "train_s2": train_s2,
    "train_s3": train_s3,
    "ground_truth": ground_truth,
    "test_s1": test_s1,
    "test_s2": test_s2,
    "test_s3": test_s3,
}

for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print("Shape:", df.shape)
    print("\nColumns:")
    print(df.columns.tolist())

    print("\nDtypes:")
    print(df.dtypes)

    print("\nMissing:")
    print(df.isna().sum())

    print("\nDuplicate rows:", df.duplicated().sum())
```

```python
gt = ground_truth.copy()

gt["matched_entity_ids"] = (
    gt["matched_entity_ids"]
    .fillna("")
    .astype(str)
)

gt["match_count"] = gt["matched_entity_ids"].apply(
    lambda x: 0 if not x.strip() else len(x.split(","))
)

print(gt["match_count"].describe())

print("\nMatch count distribution:")
print(gt["match_count"].value_counts().sort_index())
```
